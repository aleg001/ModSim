## Geomtrica

```python
geometric_samples = np.random.geometric(p=0.2, size=500)
bin = np.arange(0,11,1)
plt.hist(geometric_samples, bins=bin, edgecolor='blue', density = True) 
plt.title("Geometric Distribution") 
plt.show()
```

En este caso, lo que se quiere demostar aquí son las posiblidades que entre un cliente a nuestra tienda, y que este sea nuestro primer cliente del día. Hasta aquí, mira cual es la probabilidad que la persona oficialmente sea cliente, la primera, la segunda, la tercera, etc. hasta llegar a la décima.

```python
mean_service_time = np.mean(geometric_samples)
expected_value = 1 / 0.2

print("Valor esperado:", expected_value)
print("Media: ", mean_service_time)
```

El valor esperado esta muy cerca a la media, a diferencia de más de 0.5. Lo que significa que tendremos aproximandente un cliente al llegar al quinto cliente en la mayoría de casos.

## Binomial Negativa

```python
negative_samples = np.random.negative_binomial(n=5, p=0.4, size=500)
bin = np.arange(0,13,1)
plt.hist(negative_samples, bins=bin, edgecolor='blue', density = True) 
plt.show()

import numpy as np
import matplotlib.pyplot as plt

r = 5
p = 0.4
sample_size = 500

service_times = np.random.negative_binomial(r, p, size=sample_size)

mean_service_time = np.mean(service_times)
variance_service_time = np.var(service_times)

print(f"Promedio de los tiempos de servicio {mean_service_time:.2f} unidades.")
print(f"Variance de los tiempos de servicios {variance_service_time:.2f} unidades.")
```

Al ver la distribución llegar a lo más que se puede, se puede admirar que si esta mostrada como distribución binomal negativa. Llega a como se debe, con más de los datos estando en la izquieda, centrados entre 6 y 8 principalmente.

```python
theoretical_mean = r * (1 - p) / p
theoretical_variance = r * (1 - p) / (p**2)

print("Theoretical Mean:", theoretical_mean)
print("Theoretical Variance:", theoretical_variance)
```

La media esta muy acercada del valor real en este caso, mientras la variancia también esta cerca, solo por diferencia de dos unidades.

## Hipergeométrica

La hipergeometirca funciona con la mayoría de los datos en el medio, y después en las colas, hay menos datos, similar a la distribución normal. Lo único que sí, es que a difernecia de la normal, hay situaciones donde no se encuentran columnas en los otros lados

Por el momento, no se puede determinar si los tiempos de procesamientos entre ciertos patrones. Sin embargo, lo que se puede determinar aquí es la varianza y la media para la cantidad de muestras, donde entre más muestras correctas hay, habra menos varianza entre los datos, el tiempo promedia de unidades también sera más alto.

```python
import numpy as np

# Replace 'unknown_distribution' with the actual distribution you want to use
data = np.random.random(size=1000)
from scipy import stats

# Fit normal distribution
params_normal = stats.norm.fit(data)

# Fit exponential distribution
params_exponential = stats.expon.fit(data)

# Fit gamma distribution
params_uniform = stats.uniform.fit(data)

import matplotlib.pyplot as plt

# Plot histogram of the data
plt.hist(data, bins='auto', alpha=0.5, density=True, label='Data')

# Plot the fitted distributions
x = np.linspace(data.min(), data.max(), 100)
plt.plot(x, stats.norm.pdf(x, *params_normal), label='Normal')
plt.plot(x, stats.expon.pdf(x, *params_exponential), label='Exponential')
plt.plot(x, stats.uniform.pdf(x, *params_uniform), label='Uniform')

plt.legend()
plt.show()


```

```python
# Perform Kolmogorov-Smirnov test for each distribution
ks_stat_normal, p_value_normal = stats.kstest(data, 'norm', args=params_normal)
ks_stat_exponential, p_value_exponential = stats.kstest(data, 'expon', args=params_exponential)
ks_stat_gamma, p_value_gamma = stats.kstest(data, 'uniform', args=params_uniform)

# Interpret the results
print("Kolmogorov-Smirnov Test Results:")
print(f"Normal distribution: KS Statistic={ks_stat_normal}, p-value={p_value_normal}")
print(f"Exponential distribution: KS Statistic={ks_stat_exponential}, p-value={p_value_exponential}")
print(f"Uniforma distribution: KS Statistic={ks_stat_gamma}, p-value={p_value_gamma}")
print("El mejor resultado es la uniforme, dando un set de datos al azar.")
```
